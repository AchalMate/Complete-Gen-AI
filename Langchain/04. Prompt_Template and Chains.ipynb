```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate

from dotenv import load_dotenv
import os
```

```python
load_dotenv()
```

```python
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V4-Flash-0731",
    task="text-generation",
    max_new_tokens=200,
    temperature=0.7,
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)
```

* repo_id → which Hugging Face model we want to use
* task → what we want the model to perform
* max_new_tokens → maximum number of tokens generated
* temperature → controls randomness

```python
model = ChatHuggingFace(llm=llm)

result = model.invoke("What is the capital of India?")
print(result.content)
```

### Chains

```python
prompt = PromptTemplate(
    input_variables=["topic"],
    template="""
You are a Data Scientist.

Explain {topic} in simple language.

Include:
1. Definition
2. How it works
3. Real-world example
"""
)

chain = prompt | model
```

```python
response = chain.invoke({ "topic": "Vector Database"})

print(response)
```

```python
response.content
```

```python
topics = [
    "RAG",
    "LangChain",
    "Embeddings"
]

for topic in topics:
    response = chain.invoke({"topic": topic})
    print("=" * 80)
    print(response)
```

```python
summary_prompt = PromptTemplate(
    input_variables=["text"],
    template="""
Summarize the following text.

Rules:
- Give 5 bullet points
- Keep the language simple
- Focus only on important information

Text:
{text}
"""
)

summary_chain = summary_prompt | model
```

```python
text = """
Artificial intelligence is a branch of computer science that focuses
on building systems capable of performing tasks that normally require
human intelligence. These tasks include learning, reasoning,
problem-solving, understanding language, recognizing images, and
making decisions.
"""
```

```python
response = summary_chain.invoke({
    "text": text
})

print(response)
```
