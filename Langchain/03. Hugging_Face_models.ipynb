```python
raw_text = '''Riya was a data analyst who loved finding patterns in chaos.
One day, her company received messy sales data from multiple countries.
Everyone struggled to make sense of it, but Riya saw it as a challenge.
She spent hours cleaning, merging, and exploring the data.
Slowly, trends began to appear where others saw only confusion.
She noticed a hidden pattern in seasonal demand that no one had identified before.
Excited, she built a simple model to predict future sales.
Her predictions turned out to be surprisingly accurate.
The business team used her insights to adjust their strategy.
Within months, sales improved significantly in key regions.
Riya’s work gained attention across the organization.
She was asked to present her findings to senior leadership.
Though nervous, she explained her process clearly and confidently.
Her journey showed that clarity often comes from persistence.
From that day, Riya became known as the person who could turn data into decisions.'''


prompt = f"""
Summarize the following text in 3 concise bullet points.
Do not add information that is not present in the text.

Text:
{raw_text}
"""
result = model.invoke(prompt)

print(result.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template="""
Summarize the following text in {number} bullet points.

Text:
{text}
""",
    input_variables=["number", "text"]
)

formatted_prompt = prompt.format(
    number=3,
    text="""
Artificial intelligence is transforming many industries by automating
repetitive tasks, analyzing large amounts of data, and supporting
decision-making. AI is being used in healthcare, finance, education,
and manufacturing.
"""
)

result = model.invoke(formatted_prompt)

print(result.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are an expert document summarization assistant."
    ),
    (
        "human",
        """
Summarize the following text in {number} bullet points:

{text}
"""
    )
])

messages = prompt.format_messages(
    number=3,
    text="""
Artificial intelligence is transforming many industries by automating
repetitive tasks, analyzing large amounts of data, and supporting
decision-making. AI is being used in healthcare, finance, education,
and manufacturing.
"""
)

result = model.invoke(messages)

print(result.content)
```

```python
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are an expert Python teacher."
    ),
    (
        "human",
        "What is a list in Python?"
    ),
    (
        "ai",
        "A list is an ordered collection of items in Python."
    ),
    (
        "human",
        "Now explain it with an example."
    )
])

messages = prompt.format_messages()

result = model.invoke(messages)

print(result.content)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')

documents = [
    "Delhi is the capital of India",
    "Kolkata is the capital of West Bengal",
    "Paris is the capital of France"
]

vector = embedding.embed_documents(documents)

print(str(vector))
```
