### Chat Completion function in OpenAI

````python
msg = "Hello, GPT! This is my first ever message to you! Hi!"
messages = [{"role": "user", "content": msg}]
messages
````

#### Creating Object for OpenAI class

````python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

response =  client.chat.completions.create(
                model="gpt-4.1-mini",
                messages=messages)
print(response)
````

````python
response.choices[0].message.content
````

#### Creating Seperate functin for Chat_completion

````python
def chat_completion(messages, model="gpt-4.1-mini"):
    response = client.chat.completions.create(
        model=model,
        messages=messages
    )
    return response.choices[0].message.content
````

### How to run the function

````python
messages = [
    {"role": "user", "content": "Explain AI in simple terms"}
]

print(chat_completion(messages))
````

### Structured Output

- Instead of getting messy text, we can force the LLM to give structured data like JSON. This is very useful for automation.
### Why Structured Output?

| Without JSON       | With JSON           |
| ------------------ | ------------------- |
| Hard to parse      | Easy to use in code |
| Inconsistent       | Standard format     |
| Not usable in apps | Directly usable     |


#### 1: Simple JSON Response

````python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Give details of a product in JSON format with name, price, and category"}
]
js = chat_completion(messages)
print(js)
````

### 2.Improved Prompt for Clean JSON

````python
messages = [
    {"role": "system", "content": "You are a JSON generator. Only return valid JSON. No extra text."},
    {"role": "user", "content": """
    Generate product details in JSON format with fields:
    name, price, category
    """}
]

print(chat_completion(messages))
````

### Parsing JSON in Python (Extract JSON block)

````python
response = chat_completion(messages)

match = re.search(r"\{.*\}", response, re.DOTALL)
if match:
    json_str = match.group()
    data = json.loads(json_str)

print(data["name"])
print(data["price"])
    
````

````python
def Parse_json(response):
    match = re.search(r"\{.*\}", response, re.DOTALL)
    if match:
        json_str = match.group()
        data = json.loads(json_str)
    return data

````

### Examples

### 1. Resume Parser

````python
resume_text = """
John Doe is a Data Scientist with 3 years experience in Python and Machine Learning.
"""

messages = [
    {"role": "system", "content": "Extract structured data in JSON format"},
    {"role": "user", "content": f"""
    Extract the following from resume:
    name, experience, skills

    Resume:
    {resume_text}
    """}
]

response = chat_completion(messages)
print(response)
````

````python
data = Parse_json(response)
data
````

### 2.Multiple Records

````python
messages = [
    {"role": "system", "content": "Return only JSON array"},
    {"role": "user", "content": """
    Give 3 products in JSON list with name and price
    """}
]

response = chat_completion(messages)
print(response)
````

````python
cleaned = response.replace("```json", "").replace("```", "").strip()
data = json.loads(cleaned)

print(data)
print(data[0]["name"])  
````
