```python
from langchain_text_splitters import CharacterTextSplitter

import warnings
warnings.filterwarnings("ignore")
```

```python
text = """
LangChain helps build LLM applications.
Text splitters divide large text into chunks.
"""

splitter = CharacterTextSplitter(
    separator="\n",
    chunk_size=2,
    chunk_overlap=1
)

chunks = splitter.split_text(text)
print("Total Chunks",len(chunks))
print(chunks)
```

* CharacterTextSplitter splits on the separator first, then groups those pieces to respect.
*  chunk_size — it will never split inside a piece between two occurrences of the separator. If the separator isn't found at all, you get back the original text as one chunk, regardless of how small chunk_size is.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter


splitter = RecursiveCharacterTextSplitter(
    chunk_size=10,
    chunk_overlap=2,
    separators=[""]   # forces character-by-character splitting, ignoring words/lines
)

chunks = splitter.split_text(text)
print("Total Chunks:", len(chunks))
for i, c in enumerate(chunks):
    print(i, repr(c), "- length:", len(c))
```
