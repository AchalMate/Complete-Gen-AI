```python
import requests

def chat_with_ollama(prompt, model="llama3.1"):
    url = "http://localhost:11434/api/generate"

    data = {
        "model": model,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=data)
    return response.json()["response"]

# Chat loop
while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    response = chat_with_ollama(user_input)
    print("Bot:", response)
```

```python
import requests
import json
import re

# -------------------------------
# 1. Basic Function (Core)
# -------------------------------
def chat_with_ollama(prompt, model="llama3.1"):
    url = "http://localhost:11434/api/generate"

    payload = {
        "model": model,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=payload)
    return response.json()["response"]


# -------------------------------
# 2. Role-based Prompt (Simulated)
# -------------------------------
def chat_with_role(prompt, role="helpful assistant"):
    full_prompt = f"You are a {role}. Answer clearly.\n\nUser: {prompt}"
    return chat_with_ollama(full_prompt)
```

```python
if __name__ == "__main__":

    print("\n--- Basic Response ---")
    print(chat_with_ollama("Explain AI in simple terms"))

    print("\n--- Role-based Response ---")
    print(chat_with_role("Explain Python", role="strict teacher"))
```
